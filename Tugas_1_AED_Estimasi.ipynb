**Tugas 1: Estimasi Titik, Interval, dan Resampling**

**Data Tinggi Badan 100 Mahasiswa (dalam cm)**

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# 0. DATA 100 SAMPEL: TINGGI BADAN 100 MAHASISWA (cm)
data = np.array([
    165, 172, 158, 168, 175, 160, 163, 170, 155, 167,
    174, 162, 169, 159, 171, 166, 164, 173, 157, 168,
    176, 161, 165, 170, 156, 167, 172, 163, 169, 160,
    174, 158, 166, 171, 164, 168, 175, 159, 162, 170,
    167, 173, 161, 165, 169, 157, 172, 164, 168, 160,
    174, 163, 166, 171, 158, 169, 175, 162, 167, 170,
    164, 172, 159, 165, 168, 161, 173, 160, 166, 171,
    157, 169, 174, 163, 167, 170, 162, 168, 175, 158,
    165, 171, 164, 169, 160, 172, 166, 173, 159, 167,
    170, 161, 168, 174, 163, 165, 171, 162, 169, 166
])

n_samp = len(data)

print("=== DAFTAR 100 DATA TINGGI BADAN MAHASISWA (cm) ===")
df_sampel = pd.DataFrame({
    'Mahasiswa Ke-': range(1, n_samp + 1),
    'Tinggi Badan (cm)': data
})
print(df_sampel.to_string(index=False))
print("-" * 55)

# 1. ESTIMASI TITIK DAN INTERVAL (Distribusi Z, Alpha = 5%)
alpha = 0.05

# A. Estimasi Titik (Point Estimate): Rata-rata sampel
point_estimate = np.mean(data)

# Standar deviasi sampel dan standar error
std_dev = np.std(data, ddof=1)
std_error = std_dev / np.sqrt(n_samp)

# Nilai Z untuk tingkat kepercayaan 95% (tingkat kesalahan alpha = 5%)
z_score = stats.norm.ppf(1 - alpha / 2)

# B. Estimasi Interval (Confidence Interval)
margin_of_error = z_score * std_error
lower_bound = point_estimate - margin_of_error
upper_bound = point_estimate + margin_of_error

print("=== 1. ESTIMASI TITIK & INTERVAL (Distribusi Z) ===")
print(f"Jumlah Sampel (n)              : {n_samp}")
print(f"Estimasi Titik (Rata-rata)     : {point_estimate:.4f} cm")
print(f"Standar Deviasi Sampel (s)     : {std_dev:.4f} cm")
print(f"Tingkat Kesalahan (Alpha)      : {alpha * 100}%")
print(f"Nilai Kritis Z (Z_alpha/2)     : {z_score:.4f}")
print(f"Margin of Error                : {margin_of_error:.4f} cm")
print(f"Estimasi Interval (95% CI)     : [{lower_bound:.4f}, {upper_bound:.4f}] cm")
print("-" * 55)

# 2. RESAMPLING: METODE JACKKNIFE
jackknife_estimates = np.zeros(n_samp)

for i in range(n_samp):
    jack_sample = np.delete(data, i)
    jackknife_estimates[i] = np.mean(jack_sample)

mean_jackknife = np.mean(jackknife_estimates)

print("=== 2A. RESAMPLING METODE JACKKNIFE ===")
print(f"Jumlah Sub-sampel Jackknife    : {n_samp}")
print(f"Rata-rata Estimasi Jackknife   : {mean_jackknife:.4f} cm")
print("-" * 55)

# 3. RESAMPLING: METODE BOOTSTRAP (1000 Iterasi)
np.random.seed(42)
n_iterations = 1000
bootstrap_estimates = np.zeros(n_iterations)

for i in range(n_iterations):
    boot_sample = np.random.choice(data, size=n_samp, replace=True)
    bootstrap_estimates[i] = np.mean(boot_sample)

mean_bootstrap = np.mean(bootstrap_estimates)

print("=== 2B. RESAMPLING METODE BOOTSTRAP ===")
print(f"Jumlah Iterasi Bootstrap       : {n_iterations}")
print(f"Rata-rata Estimasi Bootstrap   : {mean_bootstrap:.4f} cm")
print("-" * 55)
```
